```python
import sys
from pathlib import Path
import pandas as pd

def find_project_root():
    path = Path.cwd().resolve()
    for candidate in [path, *path.parents]:
        if (candidate / "src" / "modeling.py").is_file():
            return candidate
    raise FileNotFoundError("Could not find src/modeling.py from " + str(path))

sys.path.insert(0, str(find_project_root() / "src"))
import modeling as m

dev = m.load("2017-2020")

for name, predictors in [("Model A (no lab)", m.MODEL_A), ("Model B (with lab)", m.MODEL_B)]:
    result = m.cross_validate(m.make_logistic, predictors, dev)
    print(f"\n=== {name} ===")
    print(m.summarise(result).round(3))
```
